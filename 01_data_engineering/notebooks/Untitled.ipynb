```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 4, figsize=(18, 5))

# Broad Gauge
axes[0].plot(df['year'], df['broad_gauge_route_km'])
axes[0].set_title('Broad Gauge Route KM')
axes[0].set_xlabel('Year')
axes[0].set_ylabel('Route KM')

# Metre Gauge
axes[1].plot(df['year'], df['metre_gauge_route_km'])
axes[1].set_title('Metre Gauge Route KM')
axes[1].set_xlabel('Year')
axes[1].set_ylabel('Route KM')

# Narrow Gauge
axes[2].plot(df['year'], df['narrow_gauge_route_km'])
axes[2].set_title('Narrow Gauge Route KM')
axes[2].set_xlabel('Year')
axes[2].set_ylabel('Route KM')

axes[3].plot(df['year'], df['total_route_km'])
axes[3].set_title('Total Route KM')
axes[3].set_xlabel('Year')
axes[3].set_ylabel('Route KM')

plt.tight_layout()
plt.show()
```

```python
df['total']=df['broad_gauge_route_km']+df['metre_gauge_route_km']+df['narrow_gauge_route_km']
```

```python
# Remove the asterisk and convert to integer
df1['Broad Gauge'] = df1['Broad Gauge'].str.replace('*', '', regex=False).astype(int)
df1['Metre Gauge'] = df1['Metre Gauge'].str.replace('*', '', regex=False).astype(int)
```

```python
# 1. Split by '-' and select the first part for every row
df1['Year'] = df1['Year'].str.split('-').str[0]

# 2. Convert the entire column to integers
df1['Year'] = df1['Year'].astype(int)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(18, 5))

# Broad Gauge
axes[0].plot(df1['Year'], df1['Broad Gauge'])
axes[0].set_title('Broad Gauge ')
axes[0].set_xlabel('Year')
axes[0].set_ylabel('Route KM')

# Metre Gauge
axes[1].plot(df1['Year'], df1['Metre Gauge'])
axes[1].set_title('Metre Gauge ')
axes[1].set_xlabel('Year')
axes[1].set_ylabel('Route KM')
```
